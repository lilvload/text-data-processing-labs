# Лабораторна робота №1

```python
import re
import nltk

from nltk.tokenize import sent_tokenize, word_tokenize, regexp_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
with open('text9.txt', 'r', encoding='utf-8') as f:
    text = f.read()

print("Початковий текст:\n")
print(text)
```

## Завдання 2
### Знайти всі номери будинків та перевернути їх у тексті (записати у зворотному порядку). Вивести змінений текст.

```python
house_regex = re.compile(
    r'(?:(?<=Street\s)|(?<=Avenue\s)|(?<=house\s)|'
    r'(?<=Road\s)|(?<=Lane\s)|(?<=Lodge\s))'
    r'(\d+(?:-[A-Za-z])?[A-Za-z]?)'      
    r'|\b(\d+[A-Za-z])\b(?!m)'          
    r'|\b(\d+-[A-Za-z])\b',            
    re.IGNORECASE
)

print("Знайдені номери будинків:")
for m in house_regex.finditer(text):
    num = m.group(1) or m.group(2) or m.group(3)
    print(f" [{m.start()}:{m.end()}] : {num} ")
```

```python
def reverse_match(match):
    full = match.group(0)
    num = match.group(1) or match.group(2) or match.group(3)
    return full.replace(num, num[::-1])

modified_text = house_regex.sub(reverse_match, text)

print("Змінений текст: \n")
print(modified_text)
```

## Завдання 3

### a) Порахувати кількість слів в тексті (не враховуючи знаки пунктуації та інші спеціальні символи) 

```python
words_only = regexp_tokenize(text, pattern=r'\w+')

print(f"Кількість слів у тексті: {len(words_only)}")
```

### б) Позначити частини мови, вивести дієслова

```python
tokens = word_tokenize(text)
pos_tagged = nltk.pos_tag(tokens)

print("Частини мови для перших 25 слів:\n")
for word, tag in pos_tagged[:25]:
    print(f"  {word:20s} {tag}")
```

```python
verbs = [word for word, tag in pos_tagged if tag.startswith('VB')]

print(f"Дієслова: \n")
print(verbs)
```

### в) Видалити стоп-слова

```python
stoplist = set(stopwords.words('english'))

cleaned_words = [w for w in words_only if w.lower() not in stoplist]

print(f"Кількість слів до видалення стоп-слів:     {len(words_only)}")
print(f"Кількість слів після видалення стоп-слів:  {len(cleaned_words)}")
print(f"Видалено:                                  {len(words_only) - len(cleaned_words)} слова "
      f"({100 * (len(words_only) - len(cleaned_words)) / len(words_only):.1f}%)")
print(f"\nПерші 30 слів після видалення:")
print(cleaned_words[:30])
```

### г) Провести пошук кореня слів п’ятого речення

```python
sentences = sent_tokenize(text)

print("Всі речення: \n")
for i, s in enumerate(sentences, 1):
    preview = s[:80] + ('...' if len(s) > 80 else '')
    print(f"  {i}. {preview}")
```

```python
fifth_sentence = sentences[4]

print("П'яте речення:\n")
print(fifth_sentence)
```

```python
pst = PorterStemmer()
fifth_tokens = regexp_tokenize(fifth_sentence, pattern=r'\w+')
stemmed = [pst.stem(w) for w in fifth_tokens]

print(f"{'Слово':<20} | {'Корінь':<20}\n")
for word, stem in zip(fifth_tokens, stemmed):
    print(f"{word:<20} | {stem:<20}")
```
